```python
datos = nba[
    (nba["Season"] == "2014 - 2015") &
    (nba["GP"] >= 50)
].copy()

datos["PPG"] = datos["PTS"] / datos["GP"]

datos.sort_values(
    "PPG", ascending=False
)[["Player", "GP", "PTS", "PPG"]].head(1)
```

```python
datos = nba[
    (nba["Season"] == "2013 - 2014") &
    (nba["3PA"] >= 100)
].copy()

datos["Porcentaje"] = (
    datos["3PM"] / datos["3PA"] * 100
)

datos.sort_values(
    "Porcentaje", ascending=False
)[["Player", "3PM", "3PA", "Porcentaje"]].head(1)
```

```python
resultado = nba.groupby("Player").agg({
    "Team": "nunique",
    "GP": "sum"
})

resultado.sort_values(
    ["Team", "GP"], ascending=False
).head(1)
```

29

```python
resultado = nba.groupby("Player")[["AST", "TOV"]].sum()

resultado = resultado[
    (resultado["AST"] >= 1000) &
    (resultado["TOV"] > 0)
].copy()

resultado["Relacion"] = (
    resultado["AST"] / resultado["TOV"]
)

resultado.sort_values(
    "Relacion", ascending=False
).head(1)
```
