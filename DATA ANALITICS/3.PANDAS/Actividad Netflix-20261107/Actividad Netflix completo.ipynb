# Actividad Dataset de Netflix

```python
import pandas as pd

netflix = pd.read_csv("netflix_titles.csv")
```

1

```python
filas, columnas = netflix.shape
print(f"Filas: {filas}")
print(f"Columnas: {columnas}")
```

2

```python
columnas = netflix.columns
print("Columnas del dataset:")
print(columnas)
```

3

```python
tipos = netflix.dtypes
print("Tipo de dato de cada columna:")
print(tipos)
```

4

```python
faltantes = netflix.isnull().sum()
print("Valores faltantes por columna:")
print(faltantes)
```

5

```python
shows_unicos = netflix["show_id"].nunique()
print(f"Shows únicos: {shows_unicos}")
```

6

```python
titulos_tipo = netflix["type"].value_counts()
print("Títulos de cada tipo:")
print(titulos_tipo)
```

7

```python
porcentaje_series = (netflix["type"] == "TV Show").mean() * 100
print(f"Porcentaje de series: {porcentaje_series:.2f}%")
```

8

```python
top5_paises = netflix["country"].value_counts().head(5)
print("Cinco países con más producciones:")
print(top5_paises)
```

9

```python
clasificacion_frecuente = netflix["rating"].value_counts().idxmax()
print(f"Clasificación más frecuente: {clasificacion_frecuente}")
```

10

```python
rating_incorrecto = netflix[netflix["rating"].str.contains("min", na=False)][["title", "rating", "duration"]]
print("Valores de rating incorrectos:")
print(rating_incorrecto)
print("En rating aparece la duración (min) y la columna duration está vacía.")
print("Seguramente al crear el dataset se desplazaron los datos una columna a la izquierda.")
```

11

```python
peliculas = netflix[netflix["type"] == "Movie"]
minutos = peliculas["duration"].str.replace(" min", "").astype(float)
peliculas_90_100 = peliculas[(minutos >= 90) & (minutos <= 100)].shape[0]
print(f"Películas entre 90 y 100 minutos: {peliculas_90_100}")
```

12

```python
peliculas_season = peliculas[peliculas["duration"].str.contains("Season", na=False)].shape[0]
print(f"Películas con 'Season' en duration: {peliculas_season}")
```

13

```python
james_brown = netflix[(netflix["type"] == "Movie") & (netflix["director"] == "James Brown")][["title"]]
print("Películas dirigidas por James Brown:")
print(james_brown)
```

14

```python
jay_karas_2016 = netflix[(netflix["type"] == "Movie") & (netflix["director"] == "Jay Karas") & (netflix["release_year"] == 2016)].shape[0]
print(f"Películas de Jay Karas publicadas en 2016: {jay_karas_2016}")
```

15

```python
top5_castings = netflix["cast"].value_counts().head(5)
print("Top 5 castings que más se repiten:")
print(top5_castings)
```

16

```python
documentales = netflix[(netflix["type"] == "Movie") & (netflix["listed_in"].str.contains("Documentaries"))]
pais_documentales = documentales["country"].value_counts().idxmax()
print(f"País con más películas de Documentaries: {pais_documentales}")
```

17

```python
stand_up = netflix[netflix["listed_in"].str.contains("Stand-Up Comedy")]
director_stand_up = stand_up["director"].value_counts().idxmax()
print(f"Director con más títulos de Stand-Up Comedy: {director_stand_up}")
```

18

```python
mas_antiguos = netflix.sort_values("release_year")[["release_year", "title"]].head(10)
print("Diez títulos más antiguos:")
print(mas_antiguos)
```

19

```python
antes_2000 = netflix[netflix["release_year"] < 2000].shape[0]
print(f"Títulos publicados antes del 2000: {antes_2000}")
```

20

```python
producciones_pais = netflix["country"].value_counts()
print("Producciones por país:")
print(producciones_pais)
```

21

```python
series = netflix[netflix["type"] == "TV Show"]
clasificacion_series = series["rating"].value_counts().idxmax()
print(f"Clasificación con más series: {clasificacion_series}")
```

22

```python
samuel_west = netflix[netflix["cast"].str.contains("Samuel West", na=False)][["title"]]
print("Títulos en los que participa Samuel West:")
print(samuel_west)
```

23

```python
director_cast = netflix.groupby(["director", "cast"]).size()
repartos_repetidos = director_cast[director_cast > 1]
print("Directores con repartos que se repiten:")
print(repartos_repetidos)
```

24

```python
vatsal_dubey = netflix[netflix["cast"].str.contains("Vatsal Dubey", na=False)]
pais_vatsal = vatsal_dubey["country"].value_counts().idxmax()
print(f"País con más títulos de Vatsal Dubey: {pais_vatsal}")
```
