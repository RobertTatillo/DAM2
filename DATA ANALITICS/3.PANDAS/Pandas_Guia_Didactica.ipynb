
## 🧰 Datasets de ejemplo (base)
Creamos diferentes datasets para reutilizar en varios bloques:
- `empleados`: personas con edad, salario, ciudad y experiencia
- `ventas`: ventas por producto y fecha
- `texto_df`: emails, nombres y comentarios


```python

empleados = pd.DataFrame({
    "id": range(1, 11),
    "nombre": ["Ana","Juan","María","Pedro","Laura","Carlos","Sofía","Miguel","Lucía","Diego"],
    "edad": [25,30,28,35,22,40,27,33,29,31],
    "salario": [30000,45000,38000,52000,28000,60000,35000,48000,41000,39000],
    "ciudad": ["Madrid","Barcelona","Madrid","Valencia","Barcelona","Madrid","Valencia","Barcelona","Sevilla","Sevilla"],
    "experiencia": [2,8,5,12,1,15,4,10,6,7]
})
empleados
```

```python

fechas = pd.date_range("2024-01-01", periods=12, freq="M")
ventas = pd.DataFrame({
    "fecha": np.repeat(fechas, 3),
    "producto": ["A","B","C"] * len(fechas),
    "unidades": np.random.randint(5, 50, size=len(fechas)*3),
    "precio_unitario": np.random.choice([9.99, 14.99, 19.99], size=len(fechas)*3)
})
ventas["importe"] = ventas["unidades"] * ventas["precio_unitario"]
ventas.head()
```

```python

texto_df = pd.DataFrame({
    "email": ["ana@gmail.com", "JUAN@hotmail.com", "maria@yahoo.es", "pedro@gmail.com", " laura@empresa.com "],
    "nombre_completo": ["Ana García López", "Juan Pérez", "María Rodríguez", "Pedro Sánchez Martín", "Laura Gómez"],
    "comentario": ["Excelente servicio!", "Muy bueno", "Regular", "Podría mejorar", "  Atención correcta  "]
})
texto_df
```

## 📊 Funciones Estadísticas


Estas funciones sirven para **describir** los datos numéricos: valores extremos, tendencia central, dispersión y acumulados.
Usaremos `empleados` y algunas columnas de `ventas`.


```python

# max / min: extremos
print("max() salario:", empleados["salario"].max())
print("min() salario:", empleados["salario"].min())

# mean / median: tendencia central
print("mean() salario:", empleados["salario"].mean())
print("median() salario:", empleados["salario"].median())

# std / var: dispersión
print("std() salario:", round(empleados["salario"].std(),2))
print("var() salario:", round(empleados["salario"].var(),2))

# sum: totales
print("sum() experiencia total:", empleados["experiencia"].sum())

# quantile: percentiles
print("quantile 25% salario:", empleados["salario"].quantile(0.25))
print("quantile 75% salario:", empleados["salario"].quantile(0.75))

# describe: resumen integral
empleados[["edad","salario","experiencia"]].describe()
```

```python

# cumsum / cumprod / cummax / cummin: acumulados ordenados por id
serie = empleados.sort_values("id")["salario"]
demo = pd.DataFrame({
    "salario": serie,
    "cumsum": serie.cumsum(),
    "cumprod(escala)": (serie/1000).cumprod().round(3),
    "cummax": serie.cummax(),
    "cummin": serie.cummin()
})
demo
```

```python

# corr / cov: relación lineal entre variables
empleados[["edad","salario","experiencia"]].corr()
```

```python

empleados[["edad","salario","experiencia"]].cov()
```

## 🔍 Selección y Filtrado


Accedemos a filas/columnas y filtramos por condiciones.


```python

# head / tail / sample
#sample(n) - Devuelve una muestra aleatoria de filas.
empleados.head(3), empleados.tail(2), empleados.sample(3, random_state=0)
```

```python

# loc: por etiquetas (índice/columnas)
empleados.loc[0:2, ["nombre","edad","salario"]]
```

```python

# iloc: por posiciones numéricas
empleados.iloc[0:3, 0:3]
```

```python

# query: expresivo y legible
empleados.query("edad > 30 and salario > 40000")
```

```python

# isin: pertenencia a lista
empleados[empleados["ciudad"].isin(["Madrid","Barcelona"])]
```

```python

# # between: rangos cerrados (incluye extremos)
empleados[empleados["edad"].between(25, 30)]
```

```python

# nlargest / nsmallest: top N por columna
#nlargest(n, col) -	Devuelve las n filas con valores más grandes.
#nsmallest(n, col) - Devuelve las n filas con valores más pequeños.
empleados.nlargest(3, "salario")[["nombre","salario"]], empleados.nsmallest(3, "salario")[["nombre","salario"]]
```

```python

# idxmax / idxmin: posición del extremo (útil para localizar fila completa)
imax = empleados["salario"].idxmax()
imin = empleados["salario"].idxmin()
empleados.loc[[imax, imin], ["nombre","salario","ciudad"]]
```

## 📝 Funciones de Texto


Operan sobre columnas de tipo `string` vía `.str`. Muy útiles en limpieza y extracción.


```python

# lower / upper
#str.lower()-Convierte todo el texto a minúsculas.
#str.upper()-Convierte todo el texto a mayúsculas.
texto_df.assign(lower=texto_df["email"].str.lower(), upper=texto_df["nombre_completo"].str.upper())
```

```python

# contains (con case-insensitive)
texto_df[texto_df["email"].str.contains("gmail")]
```

```python

# startswith / endswith
texto_df[texto_df["nombre_completo"].str.startswith("M")], texto_df[texto_df["email"].str.endswith(".com")]
```

```python

# len
texto_df["nombre_completo"].str.len()
```

```python

# split en 2 columnas: usuario y dominio bruto
texto_df["email"].str.strip().str.split("@", expand=True).rename(columns={0:"usuario",1:"dominio"})
```

```python

# replace: normalizar opiniones
# regex = false , busca exactamente “bueno”, sin interpretar símbolos especiales como . o *
texto_df
texto_df["comentario"].str.replace("bueno","excelente", regex=False)

```

```python

# strip: quitar espacios a izquierda y derecha
texto_df["email"].str.strip()
```

```python

# extract: extraer dominio base (palabras minúsculas hasta el punto)
texto_df["email"].str.strip().str.extract(r"@([a-z]+)\.", expand=False)
```

## 📊 Agregación y Agrupación

```python

# groupby + mean
empleados.groupby("ciudad")["salario"].mean().round(2)
```

```python

# agg con varias funciones
empleados.groupby("ciudad")["salario"].agg(["mean","min","max","count"]).round(2)
```

```python

# pivot_table: por defecto agrupa con mean
pd.pivot_table(ventas, values="importe", index="producto", columns=ventas["fecha"].dt.year, aggfunc=["mean","count"]).round(2)
```

```python

# crosstab: frecuencias cruzadas
#crosstab() - Crea una tabla de frecuencias cruzadas (filas vs columnas).
empleados["rango_edad"] = pd.cut(empleados["edad"], bins=[0, 30, 40, 100], labels=["Joven","Adulto","Mayor"])
pd.crosstab(empleados["ciudad"], empleados["rango_edad"])
```

## 🧹 Limpieza de Datos

```python

sucio = pd.DataFrame({
    "A": [1, 2, np.nan, 4, 5, 5, 6],
    "B": [np.nan, 2, 3, 3, np.nan, 6, 7],
    "C": ["a", "b", "c", "c", "d", "d", "e"],
    "D": [10, 20, 30, 40, 50, 60, 70]
})
sucio
```

```python

# isna / notna
sucio.isna(), sucio.notna()
```

```python

# dropna: eliminar filas con nulos; o columnas con nulos
sucio.dropna(), sucio.dropna(axis=1)
```

```python

# fillna: varios métodos
sucio.fillna(0), sucio.fillna(sucio.mean(numeric_only=True)), sucio.fillna(method="ffill")
```

```python

# drop_duplicates / duplicated
sucio.drop_duplicates(subset=["A"]), sucio.duplicated(subset=["C"]), sucio[sucio.duplicated(subset=["C"], keep=False)]
```

```python

# replace: mapeos simples o múltiples
sucio["C"].replace({"a":"A","b":"B"}), sucio.replace([1,2],[100,200])
```

```python

# interpolate: estimar valores en serie numérica
serie = pd.Series([1, np.nan, np.nan, 4, 5, np.nan, 7])
serie, serie.interpolate()
```

## 🔧 Transformación

```python

# apply: columna derivada
empleados.assign(salario_mensual = empleados["salario"] / 12).head()
```

```python

# map: a partir de diccionario
regiones = {"Madrid":"Centro", "Barcelona":"Noreste", "Valencia":"Este", "Sevilla":"Sur"}
empleados.assign(region = empleados["ciudad"].map(regiones)).head()
```

```python

# astype: conversión de tipos
empleados.assign(edad_float = empleados["edad"].astype(float)).dtypes
```

```python

# cut: categorizar por rangos
empleados.assign(cat_salario = pd.cut(empleados["salario"], bins=[0,35000,45000,1e6], labels=["Bajo","Medio","Alto"]))[[ "nombre","salario","cat_salario"]]
```

```python

# qcut: categorizar por cuantiles (mismo nº de elementos por grupo)
empleados.assign(q_salario = pd.qcut(empleados["salario"], q=4, labels=["Q1","Q2","Q3","Q4"]))[[ "nombre","salario","q_salario"]]
```

```python

# rename: renombrar columnas
empleados.rename(columns={"nombre":"name","edad":"age","ciudad":"city"}).head(2)
```

```python

# reset_index / set_index / transpose
tmp = empleados[empleados["edad"]>30]
tmp_before = tmp.index
tmp_after = tmp.reset_index(drop=True).index
(
    tmp_before.tolist(), tmp_after.tolist(),
    empleados.set_index("nombre").head(2),
    empleados[["nombre","edad","salario"]].head(3).T
)
```

## 🔗 Combinación de DataFrames

```python

left = pd.DataFrame({"id":[1,2,3,4],"nombre":["Ana","Juan","María","Pedro"],"edad":[25,30,28,35]})
right = pd.DataFrame({"id":[2,3,4,5],"ciudad":["Barcelona","Madrid","Valencia","Sevilla"],"salario":[45000,38000,52000,40000]})
left, right
```

```python

# merge: inner, left, outer
pd.merge(left, right, on="id", how="inner"), pd.merge(left, right, on="id", how="left"), pd.merge(left, right, on="id", how="outer")
```

```python

# concat: vertical y horizontal
a = pd.DataFrame({"A":[1,2], "B":[3,4]})
b = pd.DataFrame({"A":[5,6], "B":[7,8]})
pd.concat([a,b], ignore_index=True), pd.concat([a,b], axis=1)
```

```python

# join: por índice
l = pd.DataFrame({"A":[1,2,3]}, index=["a","b","c"])
r = pd.DataFrame({"B":[4,5,6]}, index=["b","c","d"])
l.join(r, how="outer")
```

```python

# append: hoy se recomienda usar concat (append está deprecado)
base = pd.DataFrame({"nombre":["Ana"],"edad":[25]})
nueva = pd.DataFrame({"nombre":["Lucía"],"edad":[26]})
pd.concat([base, nueva], ignore_index=True)
```
