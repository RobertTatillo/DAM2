#  Ejercicios de Práctica – Pandas con LaLiga 



```python
import pandas as pd
import numpy as np

laliga = pd.read_csv("laliga_player_stats_english.csv")
laliga["Minutes played"] = laliga["Minutes played"].mask(laliga["Minutes played"] % 1 != 0, laliga["Minutes played"] * 1000).round()
print(f"Dataset cargado: {laliga.shape}")
```

1. Muestra todos los jugadores del **Real Madrid**.
2. Muestra todos los jugadores del **FC Barcelona**.
3. Jugadores que hayan marcado **más de 5 goles** (`Goals scored`).
4. Jugadores que tengan **al menos 3 asistencias** (`Assists`).
5. Jugadores que hayan jugado **más de 1500 minutos** (`Minutes played`).
6. Jugadores que tengan **0 goles anotados** (`Goals scored` = 0).
7. Jugadores con **más de 30 tiros** (`Shots`).
8. Jugadores que hayan sido amonestados con **al menos 1 tarjeta roja** (`Red Cards`).


```python
# Tu código aquí

jugadores_madrid = laliga[laliga["Team"] == "Real Madrid"][["Name"]]
print("Jugadores del Real Madrid:")
print(jugadores_madrid)

jugadores_barcelona = laliga[laliga["Team"] == "FC Barcelona"][["Name"]]
print("Jugadores del FC Barcelona:")
print(jugadores_barcelona)

mas_5_goles = laliga[laliga["Goals scored"] > 5][["Name", "Goals scored"]]
print("Jugadores con más de 5 goles:")
print(mas_5_goles)

min_3_asistencias = laliga[laliga["Assists"] >= 3][["Name", "Assists"]]
print("Jugadores con al menos 3 asistencias:")
print(min_3_asistencias)

mas_1500_minutos = laliga[laliga["Minutes played"] > 1500][["Name", "Minutes played"]]
print("Jugadores con más de 1500 minutos:")
print(mas_1500_minutos)

cero_goles = laliga[laliga["Goals scored"] == 0][["Name", "Goals scored"]]
print("Jugadores con 0 goles:")
print(cero_goles)

mas_30_tiros = laliga[laliga["Shots"] > 30][["Name", "Shots"]]
print("Jugadores con más de 30 tiros:")
print(mas_30_tiros)

tarjeta_roja = laliga[laliga["Red Cards"] >= 1][["Name", "Red Cards"]]
print("Jugadores con al menos 1 tarjeta roja:")
print(tarjeta_roja)
```

## **Ejercicio 2 – Filtrado con múltiples condiciones**
Combina varias condiciones lógicas:

1. Jugadores que tengan **más de 5 goles** (`Goals scored`) **y menos de 3 asistencias** (`Assists`).
2. Jugadores que hayan jugado **más de 2000 minutos** (`Minutes played`) **y tengan al menos 1 tarjeta roja** (`Red Cards`).
3. Jugadores del **Sevilla FC** con **más de 10 tiros** (`Shots`).
4. Jugadores con **0 goles** (`Goals scored` = 0) pero **más de 1200 minutos jugados** (`Minutes played`).
5. Jugadores que hayan marcado **al menos 1 gol con la izquierda** (`Goals with left foot`) **y al menos 1 con la derecha** (`Goals with right foot`).
6. Jugadores que tengan **más de 2 goles de cabeza** (`Goals scored with header`) **y más de 20 duelos aéreos** (`Aerial duels`).


```python
# Tu código aquí

goles_pocas_asistencias = laliga[(laliga["Goals scored"] > 5) & (laliga["Assists"] < 3)][["Name", "Goals scored", "Assists"]]
print("Más de 5 goles y menos de 3 asistencias:")
print(goles_pocas_asistencias)

minutos_roja = laliga[(laliga["Minutes played"] > 2000) & (laliga["Red Cards"] >= 1)][["Name", "Minutes played", "Red Cards"]]
print("Más de 2000 minutos y al menos 1 tarjeta roja:")
print(minutos_roja)

sevilla_tiros = laliga[(laliga["Team"] == "Sevilla FC") & (laliga["Shots"] > 10)][["Name", "Shots"]]
print("Jugadores del Sevilla FC con más de 10 tiros:")
print(sevilla_tiros)

cero_goles_minutos = laliga[(laliga["Goals scored"] == 0) & (laliga["Minutes played"] > 1200)][["Name", "Goals scored", "Minutes played"]]
print("0 goles y más de 1200 minutos:")
print(cero_goles_minutos)

ambas_piernas = laliga[(laliga["Goals with left foot"] >= 1) & (laliga["Goals with right foot"] >= 1)][["Name", "Goals with left foot", "Goals with right foot"]]
print("Al menos 1 gol con la izquierda y 1 con la derecha:")
print(ambas_piernas)

cabeza_aereos = laliga[(laliga["Goals scored with header"] > 2) & (laliga["Aerial duels"] > 20)][["Name", "Goals scored with header", "Aerial duels"]]
print("Más de 2 goles de cabeza y más de 20 duelos aéreos:")
print(cabeza_aereos)
```

## **Ejercicio 3 – Agrupaciones (`groupby`)**
Trabaja con agregaciones por equipo y posición:

1. Calcula el **total de goles anotados por cada equipo** usando `Goals scored`.
2. Calcula el **total de asistencias por equipo** (`Assists`).
3. Calcula la **media de minutos jugados por posición** (`Position`, `Minutes played`).
4. Para cada equipo (`Team`), muestra el **total de tarjetas amarillas** (`Yellow Cards`) y **tarjetas rojas** (`Red Cards`).
5. Calcula el **número total de duelos ganados** (`Successful duels`) por equipo.
6. Para cada posición (`Position`), calcula la **media de tiros a puerta** (`Shots on target`).


```python
# Tu código aquí

goles_equipo = laliga.groupby("Team")["Goals scored"].sum()
print("Goles por equipo:")
print(goles_equipo)

asistencias_equipo = laliga.groupby("Team")["Assists"].sum()
print("Asistencias por equipo:")
print(asistencias_equipo)

minutos_posicion = laliga.groupby("Position")["Minutes played"].mean().round(2)
print("Media de minutos por posición:")
print(minutos_posicion)

tarjetas_equipo = laliga.groupby("Team")[["Yellow Cards", "Red Cards"]].sum()
print("Tarjetas amarillas y rojas por equipo:")
print(tarjetas_equipo)

duelos_equipo = laliga.groupby("Team")["Successful duels"].sum()
print("Duelos ganados por equipo:")
print(duelos_equipo)

tiros_puerta_posicion = laliga.groupby("Position")["Shots on target"].mean().round(2)
print("Media de tiros a puerta por posición:")
print(tiros_puerta_posicion)
```

## **Ejercicio 4 – Ordenamiento (`sort_values`)**
Ordena el DataFrame según distintas métricas:

1. Muestra los **10 máximos goleadores** ordenados de mayor a menor por `Goals scored` (muestra `Name`, `Team`, `Goals scored`).
2. Muestra los **10 jugadores con más asistencias** (`Assists`).
3. Ordena los **equipos por goles totales** (usa un `groupby` por `Team` y luego ordénalo).
4. Ordena a los jugadores por **número de tiros totales** (`Shots`) de mayor a menor.
5. Ordena a los jugadores por **minutos jugados** (`Minutes played`) de mayor a menor.
6. Muestra los **10 jugadores con más tarjetas amarillas** (`Yellow Cards`).


```python
# Tu código aquí

top10_goleadores = laliga.sort_values("Goals scored", ascending=False)[["Name", "Team", "Goals scored"]].head(10)
print("10 máximos goleadores:")
print(top10_goleadores)

top10_asistentes = laliga.sort_values("Assists", ascending=False)[["Name", "Assists"]].head(10)
print("10 jugadores con más asistencias:")
print(top10_asistentes)

equipos_goles = laliga.groupby("Team")["Goals scored"].sum().sort_values(ascending=False)
print("Equipos ordenados por goles:")
print(equipos_goles)

jugadores_tiros = laliga.sort_values("Shots", ascending=False)[["Name", "Shots"]]
print("Jugadores ordenados por tiros:")
print(jugadores_tiros)

jugadores_minutos = laliga.sort_values("Minutes played", ascending=False)[["Name", "Minutes played"]]
print("Jugadores ordenados por minutos:")
print(jugadores_minutos)

top10_amarillas = laliga.sort_values("Yellow Cards", ascending=False)[["Name", "Yellow Cards"]].head(10)
print("10 jugadores con más tarjetas amarillas:")
print(top10_amarillas)
```

## **Ejercicio 5 – Crear nuevas columnas (`assign`, operaciones)**
Crea columnas derivadas a partir de las existentes:

1. Crea una columna `Goles_por_partido` = `Goals scored` / `Games played`.
2. Crea una columna `Asistencias_por_partido` = `Assists` / `Games played`.
3. Crea una columna `Precision_tiro` = `Shots on target` / `Shots`.
4. Crea una columna `Tarjetas_totales` = `Yellow Cards` + `Red Cards`.
5. Crea una columna `Participaciones_ofensivas` = `Goals scored` + `Assists`.
6. Muestra los **15 jugadores con mayor `Goles_por_partido`** (filtra para evitar divisiones por 0 si hace falta).
7. Muestra los **15 jugadores con mejor `Precision_tiro`** (solo donde `Shots` sea mayor que 0).


```python
# Tu código aquí

laliga["Goles_por_partido"] = laliga["Goals scored"] / laliga["Games played"]

laliga["Asistencias_por_partido"] = laliga["Assists"] / laliga["Games played"]

laliga["Precision_tiro"] = laliga["Shots on target"] / laliga["Shots"]

laliga["Tarjetas_totales"] = laliga["Yellow Cards"] + laliga["Red Cards"]

laliga["Participaciones_ofensivas"] = laliga["Goals scored"] + laliga["Assists"]

print("Nuevas columnas:")
print(laliga[["Goles_por_partido", "Asistencias_por_partido", "Precision_tiro", "Tarjetas_totales", "Participaciones_ofensivas"]].head())

top15_goles_partido = laliga[laliga["Games played"] > 0].sort_values("Goles_por_partido", ascending=False)[["Name", "Goles_por_partido"]].head(15)
print("15 jugadores con más goles por partido:")
print(top15_goles_partido)

top15_precision = laliga[laliga["Shots"] > 0].sort_values("Precision_tiro", ascending=False)[["Name", "Precision_tiro"]].head(15)
print("15 jugadores con mejor precisión de tiro:")
print(top15_precision)
```

## **Ejercicio 6 – Segmentación (`pd.cut`)**
Usa `pd.cut` para crear categorías a partir de columnas numéricas:

1. Crea una nueva columna `Rango_minutos` a partir de `Minutes played` con estas categorías:
   - `Poco`: menos de 500 minutos
   - `Medio`: entre 500 y 1500 minutos
   - `Mucho`: más de 1500 minutos
   Después, muestra cuántos jugadores hay en cada categoría (`value_counts`).

2. Crea una columna `Rango_goles` a partir de `Goals scored` con estas categorías:
   - `Sin gol`: 0 goles
   - `Pocos`: 1–3 goles
   - `Goleadores`: 4 o más goles
   Muestra también cuántos jugadores hay en cada categoría.

3. Crea una columna `Rango_tiros` a partir de `Shots` con intervalos que tú elijas (por ejemplo: 0–10, 11–30, 31+). Muestra la distribución con `value_counts()`.
4. Muestra ejemplos (5 filas) de cada una de las categorías de `Rango_minutos` y `Rango_goles`.


```python
# Tu código aquí

laliga["Rango_minutos"] = pd.cut(laliga["Minutes played"], bins=[-1, 499, 1500, 10000], labels=["Poco", "Medio", "Mucho"])
jugadores_rango_minutos = laliga["Rango_minutos"].value_counts()
print("Jugadores por rango de minutos:")
print(jugadores_rango_minutos)

laliga["Rango_goles"] = pd.cut(laliga["Goals scored"], bins=[-1, 0, 3, 100], labels=["Sin gol", "Pocos", "Goleadores"])
jugadores_rango_goles = laliga["Rango_goles"].value_counts()
print("Jugadores por rango de goles:")
print(jugadores_rango_goles)

laliga["Rango_tiros"] = pd.cut(laliga["Shots"], bins=[-1, 10, 30, 1000], labels=["0-10", "11-30", "31+"])
jugadores_rango_tiros = laliga["Rango_tiros"].value_counts()
print("Jugadores por rango de tiros:")
print(jugadores_rango_tiros)

print("Ejemplos Poco:")
print(laliga[laliga["Rango_minutos"] == "Poco"][["Name", "Minutes played", "Rango_minutos"]].head(5))
print("Ejemplos Medio:")
print(laliga[laliga["Rango_minutos"] == "Medio"][["Name", "Minutes played", "Rango_minutos"]].head(5))
print("Ejemplos Mucho:")
print(laliga[laliga["Rango_minutos"] == "Mucho"][["Name", "Minutes played", "Rango_minutos"]].head(5))

print("Ejemplos Sin gol:")
print(laliga[laliga["Rango_goles"] == "Sin gol"][["Name", "Goals scored", "Rango_goles"]].head(5))
print("Ejemplos Pocos:")
print(laliga[laliga["Rango_goles"] == "Pocos"][["Name", "Goals scored", "Rango_goles"]].head(5))
print("Ejemplos Goleadores:")
print(laliga[laliga["Rango_goles"] == "Goleadores"][["Name", "Goals scored", "Rango_goles"]].head(5))
```

## **Ejercicio 7 – Tablas pivote (`pivot_table`)**
Crea resúmenes en forma de tabla:

1. Crea una tabla pivote que muestre la **media de goles y asistencias por posición** (`Position`) usando `Goals scored` y `Assists`.
2. Crea una tabla pivote con el **total de tiros a puerta (`Shots on target`) por equipo (`Team`)**.
3. Crea una tabla pivote que muestre la **media de minutos jugados (`Minutes played`) por equipo y posición** (índices: `Team`, columnas: `Position`).
4. Crea una tabla pivote con el **total de tarjetas amarillas y rojas por equipo** (`Yellow Cards`, `Red Cards`).
5. Crea una tabla pivote que muestre la **media de goles con la izquierda y la derecha por posición** (`Goals with left foot`, `Goals with right foot`, índice: `Position`).


```python
# Tu código aquí

tabla_goles_asistencias = laliga.pivot_table(index="Position", values=["Goals scored", "Assists"]).round(2)
print("Media de goles y asistencias por posición:")
print(tabla_goles_asistencias)

tabla_tiros_puerta = laliga.groupby("Team")["Shots on target"].sum()
print("Total de tiros a puerta por equipo:")
print(tabla_tiros_puerta)

tabla_minutos = laliga.pivot_table(index="Team", columns="Position", values="Minutes played").round(2)
print("Media de minutos por equipo y posición:")
print(tabla_minutos)

tabla_tarjetas = laliga.groupby("Team")[["Yellow Cards", "Red Cards"]].sum()
print("Total de tarjetas por equipo:")
print(tabla_tarjetas)

tabla_piernas = laliga.pivot_table(index="Position", values=["Goals with left foot", "Goals with right foot"]).round(2)
print("Media de goles con la izquierda y la derecha por posición:")
print(tabla_piernas)
```

## **Ejercicio 8 – Valores nulos (`isna`, `notna`)**
Analiza la calidad de los datos:

1. Muestra cuántos valores faltantes hay en cada columna (`laliga.isna().sum()`).
2. Calcula el **porcentaje de valores nulos** por columna.
3. ¿Cuántos jugadores tienen **al menos un valor nulo** en alguna columna? (pista: `.any(axis=1)`).
4. Muestra las **primeras 10 filas que contienen al menos un valor nulo**.
5. Identifica qué columnas tienen **más valores nulos**.
6. Crea un DataFrame filtrado con solo los jugadores **sin ningún valor nulo** y muestra su forma (`shape`).


```python
# Tu código aquí

nulos = laliga.isna().sum()
print("Valores nulos por columna:")
print(nulos)

porcentaje_nulos = (laliga.isna().sum() / len(laliga) * 100).round(2)
print("Porcentaje de nulos por columna:")
print(porcentaje_nulos)

jugadores_con_nulos = laliga.isna().any(axis=1).sum()
print("Jugadores con al menos un valor nulo:")
print(jugadores_con_nulos)

filas_nulos = laliga[laliga.isna().any(axis=1)].head(10)
print("Primeras 10 filas con algún nulo:")
print(filas_nulos)

columnas_mas_nulos = nulos.sort_values(ascending=False).head()
print("Columnas con más valores nulos:")
print(columnas_mas_nulos)

jugadores_sin_nulos = laliga.dropna()
print("Forma del DataFrame sin nulos:")
print(jugadores_sin_nulos.shape)
```

## **Ejercicio 9 – Análisis avanzado**
Combina lo que has aprendido para responder preguntas más complejas:

1. ¿Qué equipo tiene el **mayor número total de goles anotados**? (usa `Goals scored` y `groupby` + `sum`).
2. ¿Qué equipo tiene el **mayor número total de tarjetas rojas** (`Red Cards`)?
3. ¿Qué posición (`Position`) tiene el **mayor promedio de goles anotados por jugador**?
4. Crea una columna `Precision_tiro` = `Shots on target` / `Shots` (solo donde `Shots` > 0) y encuentra al jugador con **mejor precisión de tiro** (muestra `Name`, `Team`, `Shots`, `Shots on target`, `Precision_tiro`).
5. Crea una columna `Impacto_ofensivo` = `Goals scored` + `Assists` + `Shots on target` y muestra los **10 jugadores con mayor impacto ofensivo**.
6. Analiza si los jugadores con **más minutos jugados** suelen tener **más tarjetas amarillas**: calcula la media de `Yellow Cards` para jugadores con más de 2000 minutos y compárala con la media de jugadores con menos de 500 minutos.


```python
# Tu código aquí

equipo_mas_goles = laliga.groupby("Team")["Goals scored"].sum().sort_values(ascending=False).head(1)
print("Equipo con más goles:")
print(equipo_mas_goles)

equipo_mas_rojas = laliga.groupby("Team")["Red Cards"].sum().sort_values(ascending=False).head(1)
print("Equipo con más tarjetas rojas:")
print(equipo_mas_rojas)

posicion_mas_goles = laliga.groupby("Position")["Goals scored"].mean().sort_values(ascending=False).head(1)
print("Posición con más goles de media por jugador:")
print(posicion_mas_goles)

datos = laliga[laliga["Shots"] > 0].copy()
datos["Precision_tiro"] = datos["Shots on target"] / datos["Shots"]
mejor_precision = datos.sort_values("Precision_tiro", ascending=False)[["Name", "Team", "Shots", "Shots on target", "Precision_tiro"]].head(1)
print("Jugador con mejor precisión de tiro:")
print(mejor_precision)

laliga["Impacto_ofensivo"] = laliga["Goals scored"] + laliga["Assists"] + laliga["Shots on target"]
top10_impacto = laliga.sort_values("Impacto_ofensivo", ascending=False)[["Name", "Impacto_ofensivo"]].head(10)
print("10 jugadores con mayor impacto ofensivo:")
print(top10_impacto)

amarillas_mas_2000 = laliga[laliga["Minutes played"] > 2000]["Yellow Cards"].mean().round(2)
amarillas_menos_500 = laliga[laliga["Minutes played"] < 500]["Yellow Cards"].mean().round(2)
print(f"Media de amarillas con más de 2000 minutos: {amarillas_mas_2000}")
print(f"Media de amarillas con menos de 500 minutos: {amarillas_menos_500}")
```
