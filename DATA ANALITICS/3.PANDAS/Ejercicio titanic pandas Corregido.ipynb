## Ejercicio 8: Análisis avanzado

Realiza un análisis más profundo:

1. ¿Cuál fue la tasa de supervivencia por clase y sexo? (groupby con múltiples columnas)
2. ¿Cuáles fueron los 5 pasajeros que pagaron las tarifas más altas?
3. ¿Cuáles fueron los 5 sobrevivientes más jóvenes?
4. ¿Qué títulos tenían mayor tasa de supervivencia? (usa la columna Title del ejercicio 6)
5. ¿Cómo se relaciona el tamaño de familia con la supervivencia?
6. Ordena las clases por tarifa promedio de mayor a menor

```python
# 1. Supervivencia por clase y sexo
tasa_clase_sexo = (
    titanic.groupby(["Pclass", "Sex"])["Survived"]
           .mean()
           .round(3)
)
tasa_clase_sexo * 100
```

```python
# 2. Top 5 tarifas más altas
top_fares = (
    titanic.assign(Fare_num=pd.to_numeric(titanic["Fare"]))
           .sort_values("Fare_num", ascending=False)
           .loc[:, ["Name", "Pclass", "Sex", "Fare"]]
           .head(5)
)
top_fares
```

```python
# 3. Top 5 sobrevivientes más jóvenes
mas_jovenes = (
    titanic.loc[titanic["Survived"] == 1]
           .sort_values("Age", ascending=True)
           .loc[:, ["Name", "Age", "Pclass", "Sex"]]
           .head(5)
)
mas_jovenes
```

```python
# 4. Supervivencia por título
tmp = titanic.copy()
tmp["Title"] = tmp["Name"].str.extract(r"([A-Za-z]+)\.", expand=False)

tasa_por_titulo = (
    tmp.groupby("Title")["Survived"]
       .mean()
       .sort_values(ascending=False)
       .round(3)
)
tasa_por_titulo
```

```python
# 5. Tamaño de familia vs supervivencia
rel_familia = (
    titanic.assign(Family_Size=titanic["SibSp"] + titanic["Parch"] + 1)
           .groupby("Family_Size")["Survived"]
           .mean()
           .round(3)
           .sort_index()
)
rel_familia
```

```python
# 6. Clases ordenadas por tarifa promedio
fare_promedio_clase = (
    titanic.assign(Fare_num=pd.to_numeric(titanic["Fare"], errors="coerce"))
           .groupby("Pclass")["Fare_num"]
           .mean()
           .sort_values(ascending=False)
           .round(2)
)
fare_promedio_clase
```

## Ejercicio 10: Proyecto Final - Informe de Supervivencia

Crea un informe completo de supervivencia del Titanic que incluya:

1. **Resumen general:**
   - Total de pasajeros
   - Total de sobrevivientes
   - Tasa de supervivencia general

2. **Análisis por demografía:**
   - Supervivencia por sexo
   - Supervivencia por clase
   - Supervivencia por grupo de edad

3. **Factores de riesgo:**
   - ¿Qué grupo tuvo mayor supervivencia?
   - ¿Qué grupo tuvo menor supervivencia?
   - Top 3 conclusiones importantes

4. **Recomendaciones:**
   - Basándote en los datos, ¿qué factores fueron más importantes para la supervivencia?

```python
print("=== INFORME DE SUPERVIVENCIA DEL TITANIC ===")
print()

# 1️ RESUMEN GENERAL
print("1. RESUMEN GENERAL:")
print("===================")

total_pasajeros = len(titanic)
total_sobrevivientes = titanic["Survived"].sum()
tasa_supervivencia = total_sobrevivientes / total_pasajeros * 100

print(f"Total de pasajeros: {total_pasajeros}")
print(f"Total de sobrevivientes: {total_sobrevivientes}")
print(f"Tasa de supervivencia general: {tasa_supervivencia:.2f}%")

# 2️ ANÁLISIS POR DEMOGRAFÍA
print("\n2. ANÁLISIS POR DEMOGRAFÍA:")
print("============================")

# Supervivencia por sexo
supervivencia_sexo = titanic.groupby("Sex")["Survived"].mean() * 100
print("\nSupervivencia por sexo:")
print(supervivencia_sexo.round(2))

# Supervivencia por clase
supervivencia_clase = titanic.groupby("Pclass")["Survived"].mean() * 100
print("\nSupervivencia por clase:")
print(supervivencia_clase.round(2))

# Crear grupos de edad
titanic["Age_Group"] = pd.cut(
    titanic["Age"],
    bins=[0, 18, 60, 100],
    labels=["Child", "Adult", "Senior"]
)
titanic["Age_Group"] = titanic["Age_Group"].cat.add_categories("Unknown").fillna("Unknown")

# Supervivencia por grupo de edad
supervivencia_edad = titanic.groupby("Age_Group")["Survived"].mean() * 100
print("\nSupervivencia por grupo de edad:")
print(supervivencia_edad.round(2))

# 3️ FACTORES DE RIESGO
print("\n3. FACTORES DE RIESGO:")
print("======================")

# Unimos las tres series en una sola
todo = pd.concat([supervivencia_sexo, supervivencia_clase, supervivencia_edad])

mayor = todo.idxmax()
menor = todo.idxmin()

print(f"→ Grupo con MAYOR supervivencia: {mayor}")
print(f"→ Grupo con MENOR supervivencia: {menor}")

print("\nTop 3 conclusiones importantes:")
print("1. Las mujeres tuvieron mucha más probabilidad de sobrevivir que los hombres.")
print("2. Los pasajeros de primera clase tuvieron más oportunidades de salvarse.")
print("3. Los niños y personas jóvenes también tuvieron una tasa de supervivencia superior.")

# 4️ RECOMENDACIONES
print("\n4. RECOMENDACIONES:")
print("===================")
print(" Dar prioridad de rescate a mujeres, niños y pasajeros de primera clase.")
print(" Aumentar la disponibilidad de botes salvavidas para todas las clases.")
print(" Reforzar protocolos de emergencia en zonas de pasajeros de segunda y tercera clase.")
```
