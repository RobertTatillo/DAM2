## Ejercicio 3: Filtrado básico

Aplica filtros para encontrar información específica:

1. ¿Cuántos pasajeros sobrevivieron? (Survived = 1)
2. ¿Cuántas mujeres había en el barco?
3. ¿Cuántos pasajeros de primera clase había?
4. Encuentra todos los pasajeros menores de 18 años
5. ¿Cuántos pasajeros pagaron más de 100 en tarifa?
6. Muestra los nombres de los primeros 5 pasajeros menores de 18 años

```python
# Tu código aquí

# 1. Pasajeros que sobrevivieron
pasajeros_sobrevivientes = (titanic['Survived'] == 1).sum()
print("Pasajeros que sobrevivieron:")
print(pasajeros_sobrevivientes)

# 2. Mujeres en el barco
mujeres_barco = (titanic['Sex'] == "female").sum()
print("Mujeres en el barco:")
print(mujeres_barco)

# 3. Pasajeros de primera clase
pasajeros_primeraclase = (titanic['Pclass'] == 1).sum()
print("Pasajeros primera clase:")
print(pasajeros_primeraclase)

# 4. Menores de 18 años
pasajeros_menos18 = (titanic['Age'] < 18).sum()
print("Pasajeros menores de 18 años:")
print(pasajeros_menos18)

# 5. Tarifa mayor a 100
titanic['Fare'] = pd.to_numeric(titanic['Fare'], errors='coerce')
tarifa_mas100 = (titanic['Fare'] > 100).sum()
print("Tarifas mayor a 100:")
print(tarifa_mas100)

# 6. Nombres de los primeros 5 menores de 18
primeros_5menores = titanic[titanic['Age'] < 18]['Name'].head(5)
print("Nombres de los primeros 5 menores de 18:")
print(primeros_5menores)
```

## Ejercicio 4: Filtrado avanzado

Combina múltiples condiciones:

1. Mujeres que sobrevivieron
2. Hombres de primera clase que no sobrevivieron
3. Pasajeros entre 20 y 40 años (inclusive) que sobrevivieron
4. Pasajeros que embarcaron en 'S' o 'C' (usa isin())
5. Pasajeros cuyo nombre contiene 'Mrs.' (usa str.contains())
6. ¿Cuál fue la tasa de supervivencia de las mujeres vs los hombres?

```python
# Tu código aquí

# 1. Mujeres que sobrevivieron
mujeres_sobreviviente = ((titanic['Sex'] == "female") & (titanic['Survived'] == 1)).sum()
print("Mujeres que sobrevivieron:")
print(mujeres_sobreviviente)

# 2. Hombres de primera clase que no sobrevivieron
hombres_1clase_nosobrevivir = ((titanic['Sex'] == "male") & (titanic['Pclass'] == 1) & (titanic['Survived'] == 0)).sum()
print("Hombres de primera clase que no sobrevivieron:")
print(hombres_1clase_nosobrevivir)

# 3. Edad 20-40 que sobrevivieron
edad_2040_sobrevivir = ((titanic['Age'] >= 20) & (titanic['Age'] <= 40) & (titanic['Survived'] == 1)).sum()
print("Edad 20-40 que sobrevivieron:")
print(edad_2040_sobrevivir)

# 4. Embarcaron en 'S' o 'C'
embarcar_SoC = titanic[titanic['Embarked'].isin(["S", "C"])]
print("Pasajeros que embarcaron en S o C:")
print(len(embarcar_SoC))

# 5. Nombres con 'Mrs.'
nombres_mrs = titanic[titanic['Name'].str.contains("Mrs.")]
print("Pasajeros con 'Mrs.' en el nombre:")
print(nombres_mrs['Name'])

# 6. Tasa de supervivencia por sexo
tasa_sexo = titanic.groupby('Sex')['Survived'].mean()
print("Tasa de supervivencia por sexo:")
print(tasa_sexo)
```

## Ejercicio 5: Estadísticas y agrupaciones

Calcula estadísticas descriptivas:

1. ¿Cuál es la edad promedio de todos los pasajeros?
2. ¿Cuál es la tarifa más alta y más baja pagada?
3. ¿Cuántos pasajeros hay de cada clase? (value_counts)
4. ¿Cuál es la edad promedio por clase? (groupby)
5. ¿Cuál es la tarifa promedio por puerto de embarque?
6. Crea una tabla que muestre la cantidad de sobrevivientes y no sobrevivientes por sexo

```python
# Tu código aquí

# 1. Edad promedio
edad_promedio = titanic['Age'].mean()
print("Edad promedio:")
print(round(edad_promedio, 2))

# 2. Tarifa máxima y mínima
tarifa_maxima = titanic['Fare'].max()
tarifa_minima = titanic['Fare'].min()
print(f"Tarifa máxima: {tarifa_maxima}")
print(f"Tarifa mínima: {tarifa_minima}")

# 3. Pasajeros por clase
pasajeros_clase = titanic['Pclass'].value_counts()
print("Pasajeros por clase:")
print(pasajeros_clase)

# 4. Edad promedio por clase
edad_clase = titanic.groupby('Pclass')['Age'].mean().round(2)
print("Edad promedio por clase:")
print(edad_clase)

# 5. Tarifa promedio por puerto
tarifa_puerto = titanic.groupby('Embarked')['Fare'].mean().round(2)
print("Tarifa promedio por puerto:")
print(tarifa_puerto)

# 6. Supervivencia por sexo
supervivencia_sexo = pd.crosstab(titanic['Sex'], titanic['Survived'])
print("Supervivencia por sexo:")
print(supervivencia_sexo)
```

## Ejercicio 6: Crear nuevas columnas

Manipula el DataFrame creando nuevas columnas:

1. Crea una columna 'Family_Size' que sume SibSp + Parch + 1
2. Crea una columna 'Age_Group' con las categorías:
   - 'Child' (< 18 años)
   - 'Adult' (18-59 años)
   - 'Senior' (>= 60 años)
   - 'Unknown' (edad faltante)
3. Crea una columna 'Fare_Category' con:
   - 'Low' (Fare < 10)
   - 'Medium' (10 <= Fare < 50)
   - 'High' (Fare >= 50)
4. Crea una columna 'Title' que extraiga el título del nombre (Mr., Mrs., Miss., etc.)
5. Muestra las primeras 10 filas con las nuevas columnas

```python
# Crear una copia para no modificar el original
titanic_trabajo = titanic.copy()

# Tu código aquí

# 1. Family_Size
titanic_trabajo['Family_Size'] = titanic_trabajo['SibSp'] + titanic_trabajo['Parch'] + 1

# 2. Age_Group
def grupo_edad(edad):
    if pd.isna(edad):
        return "Unknown"
    elif edad < 18:
        return "Child"
    elif edad < 60:
        return "Adult"
    else:
        return "Senior"

titanic_trabajo['Age_Group'] = titanic_trabajo['Age'].apply(grupo_edad)

# 3. Fare_Category
def categoria_tarifa(tarifa):
    if pd.isna(tarifa):
        return "Unknown"
    elif tarifa < 10:
        return "Low"
    elif tarifa < 50:
        return "Medium"
    else:
        return "High"

titanic_trabajo['Fare_Category'] = titanic_trabajo['Fare'].apply(categoria_tarifa)

# 4. Title (pista: usa str.extract con regex)
titanic_trabajo['Title'] = titanic_trabajo['Name'].str.extract(r" ([A-Za-z]+)\.", expand=False)

# 5. Mostrar primeras 10 filas con nuevas columnas
print("Primeras 10 filas con las nuevas columnas:")
print(titanic_trabajo[['Name', 'Family_Size', 'Age_Group', 'Fare_Category', 'Title']].head(10))
```

## Ejercicio 7: Análisis de valores faltantes

Analiza los datos faltantes en el dataset:

1. ¿Qué columnas tienen valores faltantes y cuántos?
2. ¿Qué porcentaje de valores faltantes tiene cada columna?
3. ¿Cuántas filas tienen al menos un valor faltante?
4. Muestra 5 ejemplos de filas que tienen edad faltante
5. ¿Hay algún patrón en los valores faltantes de edad por clase?

```python
# Tu código aquí

# 1. Valores faltantes por columna
faltantes = titanic.isna().sum()
print("Valores faltantes por columna:")
print(faltantes[faltantes > 0])

# 2. Porcentaje de valores faltantes
porcentaje_faltantes = (titanic.isna().sum() / len(titanic) * 100).round(2)
print("Porcentaje de valores faltantes:")
print(porcentaje_faltantes)

# 3. Filas con al menos un valor faltante
filas_faltantes = len(titanic) - len(titanic.dropna())
print("Filas con al menos un valor faltante:")
print(filas_faltantes)

# 4. Ejemplos de filas con edad faltante
edad_faltante = titanic[titanic['Age'].isna()].head(5)
print("Filas con edad faltante:")
print(edad_faltante)

# 5. Patrón de edad faltante por clase
edad_faltante_clase = titanic[titanic['Age'].isna()]['Pclass'].value_counts()
print("Edad faltante por clase:")
print(edad_faltante_clase)
```

Realiza un análisis más profundo:

1. ¿Cuál fue la tasa de supervivencia por clase y sexo? (groupby con múltiples columnas)

Crea un informe completo de supervivencia del Titanic que incluya:

```python
# Tu código aquí - Desarrolla el informe completo

print("=== INFORME DE SUPERVIVENCIA DEL TITANIC ===")
print()

# 1. RESUMEN GENERAL
print("1. RESUMEN GENERAL:")
print("===================")
total_pasajeros = len(titanic)
total_sobrevivientes = titanic['Survived'].sum()
tasa_general = round(titanic['Survived'].mean() * 100, 2)
print(f"Total de pasajeros: {total_pasajeros}")
print(f"Total de sobrevivientes: {total_sobrevivientes}")
print(f"Tasa de supervivencia general: {tasa_general}%")

# 2. ANÁLISIS POR DEMOGRAFÍA
print("\n2. ANÁLISIS POR DEMOGRAFÍA:")
print("============================")
tasa_por_sexo = (titanic.groupby('Sex')['Survived'].mean() * 100).round(2)
tasa_por_clase = (titanic.groupby('Pclass')['Survived'].mean() * 100).round(2)
tasa_por_edad = (titanic_trabajo.groupby('Age_Group')['Survived'].mean() * 100).round(2)
print("Supervivencia por sexo (%):")
print(tasa_por_sexo)
print("Supervivencia por clase (%):")
print(tasa_por_clase)
print("Supervivencia por grupo de edad (%):")
print(tasa_por_edad)

# 3. FACTORES DE RIESGO
print("\n3. FACTORES DE RIESGO:")
print("======================")
tasa_grupos = (titanic.groupby(['Pclass', 'Sex'])['Survived'].mean() * 100).round(2)
grupo_mayor = tasa_grupos.idxmax()
grupo_menor = tasa_grupos.idxmin()
print(f"Grupo con mayor supervivencia: clase {grupo_mayor[0]}, {grupo_mayor[1]} ({tasa_grupos.max()}%)")
print(f"Grupo con menor supervivencia: clase {grupo_menor[0]}, {grupo_menor[1]} ({tasa_grupos.min()}%)")
print("Top 3 conclusiones:")
print(f"1. Las mujeres sobrevivieron mucho más que los hombres ({tasa_por_sexo['female']}% frente a {tasa_por_sexo['male']}%)")
print(f"2. La primera clase tuvo más supervivencia que la tercera ({tasa_por_clase[1]}% frente a {tasa_por_clase[3]}%)")
print(f"3. Los niños sobrevivieron más que los adultos ({tasa_por_edad['Child']}% frente a {tasa_por_edad['Adult']}%)")

# 4. RECOMENDACIONES
print("\n4. RECOMENDACIONES:")
print("===================")
print("El factor más importante fue el sexo, después la clase y por último la edad.")
print("Ser mujer, viajar en primera clase o ser niño aumentaba mucho las posibilidades de sobrevivir.")
```
