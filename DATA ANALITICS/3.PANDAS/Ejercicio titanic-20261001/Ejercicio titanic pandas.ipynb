## Ejercicio 2: Selección de datos

Practica la selección de columnas y filas:

1. Selecciona solo la columna 'Name' (como Series)
2. Selecciona las columnas 'Name', 'Age' y 'Sex' (como DataFrame)
3. Usando iloc, selecciona las filas 10 a 15 y las primeras 5 columnas
4. Usando loc, selecciona las filas 0 a 9 y las columnas 'Name', 'Age', 'Survived'
5. Selecciona la fila con índice 100

```python
# Tu código aquí

# 1. Columna 'Name' como Series
columna_name = titanic['Name']
print("Columna 'Name':")
print(columna_name)

# 2. Múltiples columnas como DataFrame
columnas_especificas = titanic[['Name', 'Age']]
print("Múltiples columnas:")
print(columnas_especificas)

# 3. iloc - filas 10-15, primeras 5 columnas
filas_columnas_iloc = titanic.iloc[10:16, 0:5]
print("Filas 10-15, primeras 5 columnas:")
print(filas_columnas_iloc)

# 4. loc - filas 0-9, columnas específicas
filas_columnas_loc = titanic.loc[0:9, columnas_especificas.columns]
print("Filas 0-9, columnas específicas:")
print(filas_columnas_loc)

# 5. Fila con índice 100
fila_indice_100 = titanic.loc[100]
print("Fila con índice 100:")
print(fila_indice_100)

```

## Ejercicio 3: Filtrado básico

```python
# Tu código aquí

# 1. Pasajeros que sobrevivieron
pasajeros_sobrevivientes = (titanic['Survived'] == 1).sum()
print("Pasajeros que sobrevivieron:")
print(pasajeros_sobrevivientes)

# 2. Mujeres en el barco
mujeres_barco = (titanic['Sex'] == "female").sum()
print("Mujeres en el barco:")
print(mujeres_barco)

# 3. Pasajeros de primera clase
pasajeros_primeraclase = (titanic['Pclass'] == 1).sum()
print("Pasajeros primera clase:")
print(pasajeros_primeraclase)

# 4. Menores de 18 años
pasajeros_menos18 = (titanic['Age'] < 18).sum()
print("Pasajeros menores de 18 años:")
print(pasajeros_menos18)

# 5. Tarifa mayor a 100
# titanic['Fare'] = pd.to_numeric(titanic['Fare'], errors = 'coerce').astype('Int64')
# tarifa_mas100 = (titanic['Fare'] > 100).sum()
# print("Tarifas mayor a 100:")
# print(tarifa_mas100)

# 6. Nombres de los primeros 5 menores de 18
primeros_5menores = titanic[titanic['Age'] < 18].head(5)
print("Nombres de los primeros 5 menores de 18")
print(primeros_5menores)
```
