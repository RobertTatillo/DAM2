# Operadores básicos de string

```python
mystring="Mi primera prueba con python"

#Para saber la longitud de un strig: incluido espacios
print(len(mystring))
```

```python
#Para saber dónde está un carácter, sólo da el número de la primera vez que aparece (siempre se cuenta desde 0):
print(mystring.index("p"))
```

```python
#Para contar el número de veces que aparece un carácter:
print(mystring.count("l"))
```

```python
#Para imprimir una parte de la cadena: [start:stop:step]
print(mystring[3:9:2])
```

```python
#Para invertir un string:
print(mystring[::-1])
```

```python
#Para pasar a mayus y minus:
print(mystring.upper())
print(mystring.lower())
```

```python
#Para saber si un string empieza o acaba con algo en concreto (la respuesta es True o False):
print(mystring.startswith("Hello"))
print(mystring.endswith("python"))
```

```python
#Para dividir las palabras de una cadena en una lista:
mystring = "Hello world!"
print(mystring.split(" "))
```

# Conditionals

```python
#Operadores de igualdad, desigualdad, mayor que, menor que…
x=2

print(x==2) #True
print(x<3) #True
print(x!=4) #True
```

```python
#Condicionales:
if x<1:
	print ("OK")
else:
	print("NOT OK")
```

# Loops

1. FOR

```python
# Para recorrer una lista, string…
primes=[2,3,5,7]
for prime in primes:
	print(prime)
```

```python
# Para imprimir los números en un rango:
for x in range (5):
	print(x)
print("")
for x in range (3,8,2):    #(start,stop,step)
	print(x)
```

2. WHILE

```python
# Para repetir una acción hasta que se completa una condición
count=0
while count<5:
	print(count)
	count += 1 # añade +1 en cada iteración
# Para salir de un bloque se puede usar break y para continuar continue. Se puede usar else: con los loops.
```

# Diccionarios

```python
# Inicializar un diccionario
phonebook = {
    "John" : 938477566,
    "Jack" : 938377264,
    "Jill" : 947662781
}
print(phonebook)
```

```python
# Acceder a un valor
for name, number in phonebook.items():
    print("Phone number of %s is %d" % (name, number))
```

```python
# Para quitar un valor
del phonebook["John"] #o phonebook.pop("John")
print(phonebook)
```
