## 1. Funciones Matemáticas Básicas

### `max()` y `min()` - Valores Extremos

````python
# Ejemplos básicos
ventas = [1200, 850, 1450, 980, 1650, 720, 1100]

print("Lista de ventas:", ventas)
print("Venta máxima:", max(ventas))
print("Venta mínima:", min(ventas))

# También funciona con múltiples argumentos
print("Máximo de 45, 23, 67:", max(45, 23, 67))

# Con strings (orden alfabético)
productos = ['laptop', 'tablet', 'smartphone', 'auriculares']
print("Producto alfabéticamente último:", max(productos))
print("Producto alfabéticamente primero:", min(productos))
````

### `sum()` - Suma Total

Función fundamental para calcular totales en datasets.

````python
# Suma básica
gastos_mensuales = [850, 1200, 950, 1100, 800, 1350]

total_gastos = sum(gastos_mensuales)
print(f"Gastos mensuales: {gastos_mensuales}")
print(f"Total gastado: €{total_gastos}")
print(f"Gasto promedio: €{total_gastos / len(gastos_mensuales):.2f}")

# sum() con valor inicial
print(f"Total + 100€ de comisión: €{sum(gastos_mensuales, 100)}")
````

### `len()` - Contar Elementos

Esencial para conocer el tamaño de tus datos.

````python
# Longitud de listas
temperaturas = [22, 25, 23, 27, 30, 28, 24, 26]
print(f"Datos de temperatura: {len(temperaturas)} días")

# Longitud de strings
codigo_producto = "LAPTOP-DELL-I5-8GB"
print(f"Código '{codigo_producto}' tiene {len(codigo_producto)} caracteres")

# Longitud de diccionarios
inventario = {'laptops': 15, 'tablets': 8, 'smartphones': 25}
print(f"Categorías en inventario: {len(inventario)}")
````

### `round()` y `abs()` - Redondeo y Valor Absoluto

````python
# round() para mostrar datos limpios
precio_con_descuento = 45.6789
print(f"Precio exacto: {precio_con_descuento}")
print(f"Precio redondeado: €{round(precio_con_descuento, 2)}")
print(f"Precio entero: €{round(precio_con_descuento)}")

# abs() para diferencias y distancias
objetivo_ventas = 1500
venta_real = 1235
diferencia = abs(objetivo_ventas - venta_real)
print(f"Diferencia con el objetivo: €{diferencia}")

# Útil con números negativos
cambio_porcentual = -12.5
print(f"Cambio absoluto: {abs(cambio_porcentual)}%")
````

## 2. Funciones de Iteración y Enumeración

### `range()` - Generar Secuencias Numéricas

````python
# range() básico - muy común en análisis
print("Primeros 5 números:", list(range(5)))
print("Del 1 al 10:", list(range(1, 11)))
print("De 2 en 2:", list(range(0, 11, 2)))
print("Hacia atrás:", list(range(10, 0, -1)))

# Uso práctico: generar IDs o índices
num_productos = 5
print("IDs de productos:")
for i in range(1, num_productos + 1):
    print(f"PROD-{i:03d}")  # :03d = 3 dígitos con ceros
````

### `enumerate()` - Índice + Valor

**¡MUY IMPORTANTE!** Esta función es clave en análisis de datos.

````python
# enumerate() te da posición + valor
ciudades = ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']

print("Con enumerate():")
for indice, ciudad in enumerate(ciudades):
    print(f"Posición {indice}: {ciudad}")

# Empezar desde otro número
print("\nEmpezando desde 1:")
for num, ciudad in enumerate(ciudades, 1):
    print(f"Ranking {num}: {ciudad}")

# Uso práctico: encontrar posiciones
temperaturas = [18, 22, 25, 23, 19, 21, 24]
print("\nDías con temperatura > 22°:")
for dia, temp in enumerate(temperaturas, 1):
    if temp > 22:
        print(f"Día {dia}: {temp}°C")
````

### `zip()` - Combinar Listas

Fundamental para trabajar con múltiples datasets relacionados.

### `dict()` - Combinar Listas en un diccionario



````python
# zip() combina listas elemento por elemento
meses = ['Enero', 'Febrero', 'Marzo', 'Abril']
ventas = [15000, 18000, 22000, 19500]
gastos = [12000, 14500, 16800, 15200]

print("Reporte mensual:")
for mes, venta, gasto in zip(meses, ventas, gastos):
    beneficio = venta - gasto
    print(f"{mes}: Ventas €{venta}, Gastos €{gasto}, Beneficio €{beneficio}")

# Crear diccionario con zip
productos = ['Laptop', 'Tablet', 'Phone']
precios = [899, 299, 599]
catalogo = dict(zip(productos, precios))
print(f"\nCatálogo: {catalogo}")
````

## 3. Funciones de Ordenación

### `sorted()` - Ordenar Sin Modificar Original

````python
# sorted() devuelve una NUEVA lista ordenada
puntuaciones = [87, 92, 78, 95, 83, 89, 76]

print("Original:", puntuaciones)
print("Ascendente:", sorted(puntuaciones))
print("Descendente:", sorted(puntuaciones, reverse=True))
print("Original sigue igual:", puntuaciones)

# Ordenar strings
ciudades = ['Madrid', 'Barcelona', 'Alicante', 'Zaragoza']
print("\nCiudades ordenadas:", sorted(ciudades))

# Ordenar por longitud (key=len) de menor a mayor
print("Por longitud:", sorted(ciudades, key=len))
````

### `reversed()` - Invertir Orden

````python
# reversed() invierte el orden
meses = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

print("Original:", meses)
print("Invertido:", list(reversed(meses)))

# Útil para mostrar datos más recientes primero
fechas_ventas = ['2024-01', '2024-02', '2024-03', '2024-04']
print("\nMás recientes primero:", list(reversed(fechas_ventas)))
````

## 4. Funciones Lógicas

### `any()` y `all()` - Evaluación Lógica

Muy útiles para validación de datos.

````python
# any() - ¿Alguno es True?
stocks = [15, 0, 8, 12, 0, 5]
tiene_stock_cero = any(stock == 0 for stock in stocks)
print(f"¿Hay productos sin stock? {tiene_stock_cero}")

# all() - ¿Todos son True?
todos_tienen_stock = all(stock > 0 for stock in stocks)
print(f"¿Todos los productos tienen stock? {todos_tienen_stock}")

# Ejemplo práctico: validar datos
precios = [12.5, 25.0, 18.7, 33.2]
precios_validos = all(precio > 0 for precio in precios)
print(f"¿Todos los precios son válidos? {precios_validos}")

# Detectar valores extremos
hay_precios_altos = any(precio > 30 for precio in precios)
print(f"¿Hay precios superiores a 30€? {hay_precios_altos}")
````

## 5. Funciones de Conversión de Tipos

### `int()`, `float()`, `str()` - Conversión de Datos

````python
# Conversiones básicas - muy común al leer datos
precio_texto = "29.99"
cantidad_texto = "15"

precio = float(precio_texto)
cantidad = int(cantidad_texto)
total = precio * cantidad

print(f"Precio: €{precio}")
print(f"Cantidad: {cantidad}")
print(f"Total: €{total}")

# Convertir a string para mostrar
total_texto = str(round(total, 2))
print(f"Total como texto: '{total_texto}'")

# type() para verificar tipos
print(f"\nTipo de precio: {type(precio)}")
print(f"Tipo de cantidad: {type(cantidad)}")
print(f"Tipo de total_texto: {type(total_texto)}")
````

## 6. Métodos Importantes de Listas

### `.count()` y `.index()` - Contar y Encontrar

````python
# count() cuenta apariciones
calificaciones = ['A', 'B', 'A', 'C', 'B', 'A', 'B', 'C', 'A']

print("Calificaciones:", calificaciones)
print(f"Estudiantes con A: {calificaciones.count('A')}")
print(f"Estudiantes con B: {calificaciones.count('B')}")
print(f"Estudiantes con C: {calificaciones.count('C')}")

# index() encuentra la primera posición
productos = ['laptop', 'tablet', 'smartphone', 'laptop', 'smartwatch']
primera_laptop = productos.index('laptop')
print(f"\nPrimera laptop en posición: {primera_laptop}")

# Buscar con condición
precios = [299, 599, 199, 899, 149]
for i, precio in enumerate(precios):
    if precio < 200:
        print(f"Primer producto barato en posición {i}: €{precio}")
        break
````

## 7. Funciones Avanzadas para Data Analytics

### Combinando Múltiples Funciones

````python
# Ejemplo real: análisis de satisfacción del cliente
puntuaciones_satisfaccion = [7, 9, 6, 8, 10, 5, 8, 9, 7, 6, 8, 9, 10, 4, 8]

print("=== ANÁLISIS DE SATISFACCIÓN ===")
print(f"Total respuestas: {len(puntuaciones_satisfaccion)}")
print(f"Puntuación mínima: {min(puntuaciones_satisfaccion)}")
print(f"Puntuación máxima: {max(puntuaciones_satisfaccion)}")
print(f"Promedio: {round(sum(puntuaciones_satisfaccion) / len(puntuaciones_satisfaccion), 2)}")

# Análisis por categorías
promotores = sum(1 for p in puntuaciones_satisfaccion if p >= 9)  # 9-10
detractores = sum(1 for p in puntuaciones_satisfaccion if p <= 6)  # 0-6
neutrales = len(puntuaciones_satisfaccion) - promotores - detractores

#¿Qué significa el 1?
#Es una comprensión de generador (1 for p in ... if ...).

#Lo que hace es:
#Recorre la lista puntuaciones_satisfaccion.
#Por cada elemento que cumple la condición (p >= 9), produce un 1.
#Luego sum(...) suma todos esos 1.

print(f"\nPromotores (9-10): {promotores}")
print(f"Neutrales (7-8): {neutrales}")
print(f"Detractores (0-6): {detractores}")

# NPS (Net Promoter Score)
nps = round(((promotores - detractores) / len(puntuaciones_satisfaccion)) * 100, 1)
print(f"NPS Score: {nps}")
````

### Función de Análisis Personalizada

````python
def analizar_datos(datos, nombre="Datos"):
    """
    Función que usa múltiples built-ins para análisis completo
    
    Args:
        datos: Lista de números para analizar
        nombre: Nombre descriptivo para el reporte (valor por defecto: "Datos")
    
    Returns:
        String con reporte completo de estadísticas
    """
    
    # VALIDACIÓN INICIAL
    # Verificamos si la lista está vacía usando 'not'
    # Una lista vacía es considerada False en Python
    if not datos:
        return "No hay datos para analizar"
    
    # ESTADÍSTICAS BÁSICAS usando funciones built-in
    n = len(datos)          # len() cuenta el número de elementos
    total = sum(datos)      # sum() suma todos los valores de la lista
    promedio = total / n    # División simple para calcular la media aritmética
    maximo = max(datos)     # max() encuentra el valor más alto
    minimo = min(datos)     # min() encuentra el valor más bajo
    
    # PREPARAR DATOS PARA MEDIANA
    # sorted() devuelve una NUEVA lista ordenada de menor a mayor
    # No modifica la lista original 'datos'
    datos_ordenados = sorted(datos)
    
    # CALCULAR MEDIANA (valor central)
    # La mediana cambia según si hay cantidad par o impar de elementos
    if n % 2 == 0:  # Si es par (el resto de dividir por 2 es 0)
        # Para cantidad par: promedio de los dos valores centrales
        # n//2 - 1 y n//2 son las posiciones de los dos elementos centrales
        mediana = (datos_ordenados[n//2 - 1] + datos_ordenados[n//2]) / 2
    else:           # Si es impar
        # Para cantidad impar: el elemento exactamente en el centro
        # n//2 nos da la posición central (división entera)
        mediana = datos_ordenados[n//2]
    
    # CALCULAR RANGO (dispersión)
    # Rango = diferencia entre el valor máximo y mínimo
    # Nos indica qué tan dispersos están los datos
    rango = maximo - minimo
    
    # GENERAR REPORTE FORMATEADO
    # Usamos f-string multilínea para crear un reporte profesional
    reporte = f"""
=== ANÁLISIS DE {nombre.upper()} ===
Muestra: {n} valores
Suma total: {round(total, 2)}
Promedio: {round(promedio, 2)}
Mediana: {round(mediana, 2)}
Mínimo: {minimo}
Máximo: {maximo}
Rango: {rango}
Datos ordenados: {datos_ordenados[:5]}{'...' if n > 5 else ''}
"""
    
    # EXPLICACIÓN DE LA ÚLTIMA LÍNEA:
    # datos_ordenados[:5] - muestra solo los primeros 5 valores (slicing)
    # {'...' if n > 5 else ''} - añade "..." solo si hay más de 5 elementos
    # Es una expresión condicional inline (ternaria)
    
    return reporte  # Devolvemos el string con todo el análisis

# PROBAR LA FUNCIÓN
# Lista de ejemplo con datos de ventas mensuales
ventas_mes = [1200, 980, 1450, 1100, 1650, 890, 1320, 1180, 1400, 1250]

# Llamamos la función y mostramos el resultado
# El segundo parámetro personaliza el título del reporte
print(analizar_datos(ventas_mes, "Ventas del Mes"))

# FUNCIONES BUILT-IN UTILIZADAS EN ESTA FUNCIÓN:
# len() - contar elementos
# sum() - sumar todos los valores
# max() - valor máximo
# min() - valor mínimo
# sorted() - ordenar sin modificar original
# round() - redondear decimales
# upper() - convertir texto a mayúsculas
````
