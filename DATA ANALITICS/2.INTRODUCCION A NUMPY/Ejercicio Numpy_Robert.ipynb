## Ejercicio 2: Operaciones con Arrays

Dados los siguientes arrays, realiza las operaciones indicadas:

```python
precios = np.array([12.50, 8.30, 15.75, 22.40, 9.80])
cantidades = np.array([3, 5, 2, 1, 4])
```

Calcula:
1. El total por producto (precio × cantidad)
2. Aplica un descuento del 10% a todos los precios
3. Encuentra qué productos cuestan más de 10 euros
4. Calcula el precio promedio
5. Encuentra el producto más caro y el más barato (usa `argmax` y `argmin`)

````python
# Datos
precios = np.array([12.50, 8.30, 15.75, 22.40, 9.80])
cantidades = np.array([3, 5, 2, 1, 4])

# Tu código aquí

# 1. Total por producto
print("Precios totales por producto: ", precios * cantidades)

# 2. Precios con 10% descuento
print("Precios con 10% de descuento: ", precios * 0.9)

# 3. Productos que cuestan más de 10€
np_mas_de_diez = precios[precios > 10]
print("Productos que cuestan más de 10€: ", np_mas_de_diez)

# 4. Precio promedio
np_promedio = np.mean(precios)
print("Precio promedio: ", np_promedio)

# 5. Producto más caro y más barato (usa argmax y argmin)
print("Producto más caro: ", np.argmax(precios))
print("Producto más barato: ", np.argmin(precios))
````

## Ejercicio 3: Indexing y Slicing

Dado el array de temperaturas de una semana:

```python

```

Responde:
1. ¿Cuál fue la temperatura del miércoles?
2. ¿Cuáles fueron las temperaturas de los primeros 3 días?
3. ¿Cuáles fueron las temperaturas del fin de semana (sábado y domingo)?
4. ¿Qué días tuvieron temperatura superior a 25°?
5. ¿Cuál es la temperatura promedio de los días laborables (lunes a viernes)?

````python
# Datos
temperaturas = np.array([18, 22, 25, 19, 27, 30, 24])
dias = np.array(['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom'])

# Tu código aquí

# 1. Temperatura del miércoles (índice 2)
temp_miercoles = temperaturas[2]
print(f"Temperatura del miércoles: {temp_miercoles}°C")

# 2. Temperaturas primeros 3 días
temp_primeros_3 = temperaturas[:3]
print(f"Temperaturas primeros 3 días: {temp_primeros_3}")

# 3. Temperaturas fin de semana
temp_fin_semana = temperaturas[5:7]
print(f"Temperaturas fin de semana: {temp_fin_semana}")

# 4. Días con temperatura > 25°
dias_alta_temp = dias[temperaturas > 25]
print(f"Días con temperatura > 25°: {dias_alta_temp}")

# 5. Promedio días laborables (índices 0-4)
temp_promedio_laborables = np.mean(temperaturas[:5])
print(f"Promedio días laborables: {temp_promedio_laborables}°C")
````

## Ejercicio 5: Arrays 2D (Matrices)

Crea una matriz que represente las ventas de 3 productos durante 4 meses:

```
           Ene  Feb  Mar  Abr
Producto A: 120  150  180  200
Producto B:  80   90  110  130
Producto C: 200  220  250  280
```

Responde:
1. Crea la matriz de ventas
2. ¿Cuáles fueron las ventas del Producto B en marzo?
3. ¿Cuáles fueron las ventas totales de febrero (todos los productos)?
4. ¿Cuál producto tuvo las mayores ventas totales en el período?
5. ¿En qué mes se vendió más en total?
6. Calcula el promedio de ventas por producto
7. Calcula el promedio de ventas por mes

````python
# Tu código aquí

# 1. Crear matriz de ventas
ventas = np.array([[120, 150, 180, 200], [80, 90, 110, 130], [200, 220, 250, 280]])
productos = np.array(["Producto A", "Producto B", "Producto C"])
meses = np.array(["Enero", "Febrero", "Marzo", "Abril"])

# 2. Ventas Producto B en marzo
ventasB = ventas[1, 2]
print(f"Ventas Producto B en marzo: {ventasB}")

# 3. Ventas totales de febrero
ventas_febrero = np.sum(ventas[:, 1])
print(f"Ventas totales de febrero: {ventas_febrero}")

# 4. Producto con mayores ventas totales
ventas_productos = np.sum(ventas, axis=1)
producto_max_ventas = np.argmax(ventas_productos)
print(f"Producto con mayores ventas totales: {productos[producto_max_ventas]}")

# 5. Mes con más ventas
ventas_meses = np.sum(ventas, axis=0)
mes_max_ventas = np.argmax(ventas_meses)
print(f"Mes con más ventas: {meses[mes_max_ventas]}")

# 6. Promedio por producto
promedio_productos = np.mean(ventas, axis=1)
print(f"Promedio por producto: {promedio_productos}")

# 7. Promedio por mes
promedio_meses = np.mean(ventas, axis=0)
medias_redondeadas = np.round(promedio_meses, 2)
print(f"Promedio por mes: {medias_redondeadas}")
````

## Ejercicio 6: Análisis de Datos de Sensores

Un sensor registra temperaturas cada hora durante un día. Los datos fueron:

```python
temperaturas_sensor = np.array([15.2, 16.1, 17.3, 19.2, 22.1, 25.4, 28.3, 30.1, 
                               31.5, 32.2, 33.1, 31.8, 30.5, 28.9, 26.7, 24.3, 
                               22.1, 20.5, 18.9, 17.2, 16.1, 15.8, 15.3, 15.1])
```

Realiza el siguiente análisis:
1. ¿Cuál fue la temperatura máxima y a qué hora ocurrió?
2. ¿Cuál fue la temperatura mínima y a qué hora ocurrió?
3. Calcula la temperatura promedio del día
4. ¿En cuántas horas la temperatura estuvo por encima de 25°?
5. Calcula la diferencia entre la temperatura máxima y mínima
6. Encuentra las horas donde la temperatura estuvo entre 20° y 25° (inclusive)

````python
# Datos
temperaturas_sensor = np.array([15.2, 16.1, 17.3, 19.2, 22.1, 25.4, 28.3, 30.1,
                               31.5, 32.2, 33.1, 31.8, 30.5, 28.9, 26.7, 24.3, 
                               22.1, 20.5, 18.9, 17.2, 16.1, 15.8, 15.3, 15.1])

# Tu código aquí

# 1. Temperatura máxima y hora
temp_max = np.max(temperaturas_sensor)
hora_max = np.argmax(temperaturas_sensor)
print(f"Temperatura máxima: {temp_max}°C a las {hora_max} horas")

# 2. Temperatura mínima y hora
temp_min = np.min(temperaturas_sensor)
hora_min = np.argmin(temperaturas_sensor)
print(f"Temperatura mínima: {temp_min}°C a las {hora_min} horas")

# 3. Temperatura promedio
temp_promedio = np.mean(temperaturas_sensor)
temp_promdio_redondeado = np.round(temp_promedio, 2)
print(f"Temperatura promedio: {temp_promdio_redondeado}°C")

# 4. Horas con temperatura > 25°
horas_alta = np.where(temperaturas_sensor > 25)[0]
print(f"Horas con temperatura > 25°: {horas_alta}")

# 5. Diferencia máxima-mínima
diferencia = temp_max - temp_min
print(f"Diferencia máxima-mínima: {diferencia}°C")

# 6. Horas con temperatura entre 20° y 25°
horas_mediana = np.where((temperaturas_sensor >= 20) & (temperaturas_sensor <= 25))[0]
print(f"Horas con temperatura entre 20° y 25°: {horas_mediana}")
````

## Ejercicio 7: Análisis Financiero

Una empresa tiene los siguientes datos trimestrales de ingresos y gastos:

```python
ingresos = np.array([50000, 65000, 70000, 80000])  # Q1, Q2, Q3, Q4
gastos = np.array([35000, 45000, 50000, 55000])   # Q1, Q2, Q3, Q4
```

Calcula:
1. El beneficio por trimestre (ingresos - gastos)
2. El beneficio total del año
3. El trimestre con mayor beneficio
4. El margen de beneficio por trimestre (beneficio/ingresos * 100)
5. ¿En qué trimestres el margen fue superior al 25%?
6. La tasa de crecimiento de ingresos trimestre a trimestre
7. El promedio de gastos del año

````python
# Datos
ingresos = np.array([50000, 65000, 70000, 80000])  # Q1, Q2, Q3, Q4
gastos = np.array([35000, 45000, 50000, 55000])   # Q1, Q2, Q3, Q4
trimestres = np.array(['Q1', 'Q2', 'Q3', 'Q4'])

# Tu código aquí

# 1. Beneficio por trimestre
beneficio_trimestre = ingresos - gastos
print(f"Beneficio por trimestre: {beneficio_trimestre}")

# 2. Beneficio total del año
beneficio_total = np.sum(beneficio_trimestre)
print(f"Beneficio total del año: {beneficio_total}")

# 3. Trimestre con mayor beneficio
trimestre_max_beneficio = np.argmax(beneficio_trimestre)
print(f"Trimestre con mayor beneficio: {trimestres[trimestre_max_beneficio]}")

# 4. Margen de beneficio por trimestre (%)
margen_beneficio = (beneficio_trimestre / ingresos) * 100
margen_beneficio_redondeado = np.round(margen_beneficio, 2)
print(f"Margen de beneficio por trimestre (%): {margen_beneficio_redondeado}")

# 5. Trimestres con margen > 25%
trimestres_altos = trimestres[margen_beneficio > 25]
print(f"Trimestres con margen > 25%: {trimestres_altos}")

# 6. Tasa de crecimiento de ingresos
tasa_crecimiento = np.diff(ingresos) / ingresos[:-1] * 100
tasa_crecimiento_redondeada = np.round(tasa_crecimiento, 2)
print(f"Tasa de crecimiento de ingresos: {tasa_crecimiento_redondeada}%")

# 7. Promedio de gastos
promedio_gastos = np.mean(gastos)
promedio_gastos_redondeado = np.round(promedio_gastos, 2)
print(f"Promedio de gastos: {promedio_gastos_redondeado}")
````
