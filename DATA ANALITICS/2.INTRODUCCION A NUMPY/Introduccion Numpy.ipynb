# Introducción a NumPy para Data Analytics

NumPy es la librería fundamental para computación científica en Python y la base de Pandas.

## 1. Importar NumPy

```python
import numpy as np
print(f"NumPy versión: {np.__version__}")
```

## 2. ¿Por qué NumPy?

Comparemos Python básico vs NumPy:

```python
# Python básico - lento
lista = [1, 2, 3, 4, 5]
lista_doble = [x * 2 for x in lista]
print("Python básico:", lista_doble)

# NumPy - rápido y simple
array = np.array([1, 2, 3, 4, 5])
array_doble = array * 2
print("NumPy:", array_doble)
print("\n¡NumPy es más simple y hasta 100x más rápido!")
```

## 3. Crear Arrays

```python
# Desde listas
arr1 = np.array([1, 2, 3, 4])
print("Desde lista:", arr1)

# Arrays especiales
zeros = np.zeros(5)
ones = np.ones(4)
rango = np.arange(0, 10, 2)

print("Zeros:", zeros)
print("Ones:", ones)
print("Rango:", rango)
```

## 4. Propiedades de Arrays

```python
# Ejemplo: notas de estudiantes
notas = np.array([85, 92, 78, 96, 88, 73, 89])

print("Notas:", notas)
print(f"Forma: {notas.shape}")     # (7,)
print(f"Tamaño: {notas.size}")     # 7
print(f"Tipo: {notas.dtype}")      # int64
print(f"Dimensiones: {notas.ndim}") # 1
```

## 5. Operaciones Básicas

```python
# Operaciones aritméticas
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("a:", a)
print("b:", b)
print("a + b:", a + b)
print("a * 3:", a * 3)
print("a ** 2:", a ** 2)

# Comparaciones
print("a > 2:", a > 2)
```

## 6. Indexing y Slicing

```python
ventas = np.array([100, 150, 80, 200, 95, 180])
print("Ventas:", ventas)

# Indexing básico
print("Primera venta:", ventas[0])
print("Última venta:", ventas[-1])
print("Primeras 3:", ventas[:3])

# Boolean indexing
ventas_altas = ventas > 150
print("¿Ventas > 150?:", ventas_altas)
print("Ventas altas:", ventas[ventas_altas])
```

## 7. Funciones Matemáticas y Estadísticas

```python
# Datos de ejemplo
datos = np.array([10, 20, 30, 40, 50])
print("Datos:", datos)

# Funciones matemáticas
print("Raíz cuadrada:", np.sqrt(datos))
print("Logaritmo:", np.log(datos))

# Estadísticas
print(f"Media: {np.mean(datos)}")
print(f"Mediana: {np.median(datos)}")
print(f"Máximo: {np.max(datos)}")
print(f"Mínimo: {np.min(datos)}")
print(f"Suma: {np.sum(datos)}")
print(f"Desviación estándar: {np.std(datos):.2f}")
```

## 8. Arrays 2D (Matrices)

```python
# Crear matriz 2D
matriz = np.array([[1, 2, 3], 
                   [4, 5, 6]])
print("Matriz:")
print(matriz)
print(f"Forma: {matriz.shape}")  # (2, 3)

# Acceso a elementos
print(f"Elemento [0,1]: {matriz[0, 1]}")
print(f"Primera fila: {matriz[0, :]}")
print(f"Segunda columna: {matriz[:, 1]}")
```

## 9. Ejemplo Práctico: Análisis de Ventas

```python
# Datos de ventas trimestrales de 3 productos
ventas_trimestre = np.array([[120, 150, 180],  # Producto A
                            [80, 90, 110],    # Producto B  
                            [200, 220, 250]]) # Producto C

productos = ['Producto A', 'Producto B', 'Producto C']
trimestres = ['Q1', 'Q2', 'Q3']

print("Ventas por trimestre:")
print("       Q1   Q2   Q3")
for i, producto in enumerate(productos):
    print(f"{producto}: {ventas_trimestre[i]}")
print()

# Análisis
print("ANÁLISIS:")
print(f"Total general: {np.sum(ventas_trimestre)}")
print(f"Promedio general: {np.mean(ventas_trimestre):.1f}")

# Por producto (filas)
ventas_por_producto = np.sum(ventas_trimestre, axis=1)
print("\nVentas totales por producto:")
for i, producto in enumerate(productos):
    print(f"{producto}: {ventas_por_producto[i]}")

# Por trimestre (columnas)
ventas_por_trimestre = np.sum(ventas_trimestre, axis=0)
print("\nVentas totales por trimestre:")
for i, trimestre in enumerate(trimestres):
    print(f"{trimestre}: {ventas_por_trimestre[i]}")

# Mejor producto y trimestre
mejor_producto = np.argmax(ventas_por_producto)
mejor_trimestre = np.argmax(ventas_por_trimestre)
print(f"\nMejor producto: {productos[mejor_producto]}")
print(f"Mejor trimestre: {trimestres[mejor_trimestre]}")
```

## 10. Ejemplo Práctico: BMI (Índice de Masa Corporal)

```python
# Datos de personas
nombres = np.array(['Ana', 'Juan', 'María', 'Carlos'])
altura = np.array([1.65, 1.78, 1.60, 1.85])  # metros
peso = np.array([58, 75, 52, 82])  # kg

print("Datos de personas:")
for i in range(len(nombres)):
    print(f"{nombres[i]}: {altura[i]}m, {peso[i]}kg")
print()

# Calcular BMI = peso / altura²
bmi = peso / (altura ** 2)
bmi = np.round(bmi, 1)

print("BMI calculado:")
for i in range(len(nombres)):
    print(f"{nombres[i]}: {bmi[i]}")
print()

# Clasificar BMI
peso_normal = (bmi >= 18.5) & (bmi < 25)
sobrepeso = bmi >= 25

print("Clasificación:")
print(f"Peso normal: {nombres[peso_normal]}")
print(f"Sobrepeso: {nombres[sobrepeso]}")
print(f"BMI promedio: {np.mean(bmi):.1f}")
```
